### .loc()
##### Mengakses data berdasarkan LABEL (nama index & kolom)
df.loc[baris, kolom]


```python
df.loc[0, 'agent']        # baris index 0, kolom 'agent'
```

```python
df.loc[:, 'children']       # semua baris, kolom 'children'
```

```python
df.loc[df['agent'].isna(), :]  # filter data
```

### .iloc()
##### Mengakses data berdasarkan POSISI (angka index)
df.iloc[baris_ke, kolom_ke]


```python
df.iloc[0, 1]     # baris pertama, kolom kedua
```

```python
df.iloc[:, 0]     # semua baris, kolom pertama
```

```python
df.iloc[5:10, :]  # slicing
```
