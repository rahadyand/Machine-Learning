# Feature Engineering
Membuat atau memodifikasi feature agar model machine learning bisa bekerja lebih baik

```python
import pandas as pd 
```

```python
df = pd.read_csv('Data/hotel_bookings.csv')
df['country'] = df['country'].fillna('Unknown') 
df['children'] = df['children'].fillna(0)
df['agent'] = df['agent'].fillna(0)
df.drop(columns=['company'], axis=1, inplace=True)
df.drop_duplicates(inplace=True)
df
```

```python
# Total malam setiap minggu
df['total_nights'] = df['stays_in_week_nights']+ df['stays_in_weekend_nights']
```

```python
df[['total_nights', 'stays_in_week_nights', 'stays_in_weekend_nights']].sample(10)
```

```python
df['customers'] = df['adults'] + df['children'] + df['babies']
df[['customers', 'adults', 'children', 'babies']].sample(10)
```

#### .apply()
Menerapkan sebuah fungsi ke setiap elemen, baris, atau kolom

---------------------

##### lambda x == def
fungsi anonim singkat yang sering dipakai di apply() 

```python
# Mengubah 0 & 1 menjadi No & Yes
df['is_repeated_category'] = df['is_repeated_guest'].apply(lambda x: 'Yes' if x == 1 else 'No' )
```

```python
df[['is_repeated_category', 'is_repeated_guest']].sample(10)
```

### Ubah tipe data object -> date time
Kenapa harus di kanan dan kiri, agar tidak membuat kolom baru hanya mengganti kolom yang dibuah

```python
df['reservation_status_date'] = pd.to_datetime(df['reservation_status_date'])
```

```python
df['reservation_status_date']
```

#### .map()
digunakan untuk memetakan (mapping) nilai lama menjadi nilai baru berdasarkan:
- dictionary
- fungsi
- lambda

#### (Boolean) -> lambda 
#### (Multiclass) -> map

```python
# Mengubah nama bulan menjadi angka
month_mapping = {
    'January': 1,
    'February': 2,
    'March': 3,
    'April': 4,
    'May': 5,
    'June': 6,
    'July': 7,
    'August': 8,
    'September': 9,
    'October': 10,
    'November': 11,
    'December': 12
}

df['arrival_date_month'] = df['arrival_date_month'].map(month_mapping)
```
