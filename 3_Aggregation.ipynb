# Aggregation
menggabungkan beberapa feature untuk melihat insight apa yang bisa didapatkan

```python
import pandas as pd 
```

```python
df = pd.read_csv('Data/hotel_bookings.csv')
df['country'] = df['country'].fillna('Unknown') 
df['children'] = df['children'].fillna(0)
df['agent'] = df['agent'].fillna(0)
df.drop(columns=['company'], axis=1, inplace=True)
df.drop_duplicates(inplace=True)
df
```

### .groupby()
Menggabungkan beberapa feature yang berkolerasi

------------------

##### .value_counts()
Mengitung jumlah

```python
# hotel & is_canceled -- Hotel mana aja yang di canceled
df.groupby('hotel')['is_canceled'].value_counts().unstack()
```

```python
# adr = average daily rate -- Mencari harga rata rata pada hotel
df.groupby('hotel')['adr'].mean()
```

```python
# Harga rata rata hotel per tahun
df.groupby(['arrival_date_year','hotel'])['adr'].mean().unstack()
```

#### .agg()
.agg() (aggregate) adalah untuk melakukan satu atau beberapa operasi agregasi pada data

```python
# Jumlah hari pemesanan setiap tamu
df.groupby('customer_type')['lead_time'].agg(['min', 'max', 'mean', 'count'])
```
