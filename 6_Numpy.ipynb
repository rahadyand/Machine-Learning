### Dot product & Matrix Multiplication
- Dot product → operasi antara dua vektor yang menghasilkan satu nilai skalar (example: 2)
- Matrix multiplication → menghitung banyak dot product sekaligus -- wajib (m x n ; n x p) hasilnya m x p

```python
v1 = np.array([1, 2, 3])
v2 = np.array([4, 5, 6])
```

```python
# np.dot = dot product
hasil_dot = np.dot(v1, v2)  
hasil_dot
```

```python
m1 = np.array([[2, 3, 1],
               [7, 6, 1]])
m2 = np.array([[1, 2],
                [5, 3],
                [1, 1]])
```

```python
hasil_multiplication = np.dot(m1, m2)
hasil_multiplication
```

```python
hasil_multiplication = np.matmul(m1, m2)
hasil_multiplication
```

### np.concatenate
mengabungkan 2 array  
axis = 0 - (rows)  
axis = 1 - (columns)  

```python
array1 = np.array([[1, 2], [3, 4]])
array2 = np.array([[5, 6]])

array_gabungan = np.concatenate((array1, array2), axis=0)
array_gabungan
```

```python
array1 = np.array([[1, 2], [3, 4]])
array2 = np.array([[5],
                   [6]])

array_gabungan = np.concatenate((array1, array2), axis=1)
array_gabungan
```
