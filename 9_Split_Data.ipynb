### Kita ingin memastikan model benar-benar "paham" polanya, bukan sekadar "menghafal" data yang diberikan.

Pilih Train-Test Split Jika:
- Data Sangat Besar (Jutaan baris). Cross-validation akan terlalu lama.
- Ingin hasil cepat (prototyping).

Pilih Cross-Validation Jika:
- Data Sedikit/Terbatas (Sayang kalau dibuang cuma buat tes).
- Data tidak seimbang (imbalanced).
- Ingin hasil yang paling valid untuk laporan ilmiah/skripsi.

```python
import pandas as pd
```

```python
df = pd.read_csv('Data/heart.csv')
df
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
model = LogisticRegression(max_iter=1000, random_state=42)
```

-----------------------------

# Train Test Split
Tujuan utama model = generalisasi  

Tujuan melakukan split = Seberapa baik model ini akan bekerja pada data baru yang belum pernah dilihat sebelumnya

- x_train = latihan soal
- y_train = kuci jawaban latihan soal

- x_test = ujian
- y_test = kunci jawaban ujian



```python
from sklearn.model_selection import train_test_split
```

```python
x = df.drop('target', axis=1)
y = df['target']
```

- textsize = 0.2 (membagi train:test 80:20)
- random_state = 42 (Mengunci randomisasi)
- stratify = y (Memastikan perbandingan label di train dan test sesuai dengan data asli) -- Balance


```python

x_train, x_test, Y_train, Y_test = train_test_split(x, y, test_size=0.2, random_state=42, stratify=y)
```

# Cross Validation
- Metode: Bagi data jadi $K$ bagian, latih & uji berulang-ulang secara bergantian.
- Sifat: Lama, tapi hasil sangat akurat & adil (robust).
- Analogi: Marathon Ujian Harian. Semua bab buku pernah jadi bahan belajar, dan semua bab pernah jadi soal ujian. Nilai diambil dari rata-rata.

------------------------------------------

- n_splits = 5 (Jumlah iterasi)
- shuffle = True ()
- random state = 42 (Mengunci randomnisasi)

```python
from sklearn.model_selection import StratifiedKFold, cross_val_score
```

```python
kf  = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
```

cv_score = melihat hasil model in general  
digunakan untuk = menjadi treshold

```python
cv_score = cross_val_score(model, x_train, Y_train, cv=kf, scoring='accuracy')
```

```python
cv_score.mean()
```

### i

```python
model.fit(x_train, Y_train)
```

```python
y_pred = model.predict(x_test)
```

```python
from sklearn.metrics import accuracy_score
```

```python
test_accuracy = accuracy_score(Y_test, y_pred) 
test_accuracy
```
