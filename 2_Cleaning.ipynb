# Data Cleaning

```python
import pandas as pd 
```

```python
df = pd.read_csv('Data/hotel_bookings.csv')
df
```

## remove missing value
### .isnull()
### .isna()

```python
df.isna().sum()
```

```python
# ubah ke presentase
df.isna().sum() / len(df) * 100
```

### Mengisi dengan Unknown
.fillna()

```python
df['country'] = df['country'].fillna('Unknown') 
```

### Mengisi dengan 0

```python
df['children'] = df['children'].fillna(0)
```

```python
df['agent'] = df['agent'].fillna(0)
```

### Menghapus kolom karena NaN yang terlalu banyak
.drop()

-- Perbedaan inplace(True) dan inplace(False)
- True = langsung mengubah data frame asli
- false/ default = tidak mengubah data frame/ data frame lama tetap utuh 

```python
# 3️⃣ Contoh Perbandingan

df1 = pd.DataFrame({
    'A': [1, 2],
    'B': [3, 4]
})

df1.drop('A', axis=1, inplace=False)
print(df1)      # Kolom A MASIH ADA

df1.drop('A', axis=1, inplace=True)
print(df1)      # Kolom A HILANG
```

-- Perbedaan axis=1 & axis=0
- 1 = kolom ('Nama Kolom')
- 0 = baris (Index)

-----

```python
dfNew = df.drop(columns=['company'], axis=1, inplace=False)
```
